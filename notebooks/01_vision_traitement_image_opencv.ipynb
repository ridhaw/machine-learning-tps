```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import cv2

file_path = '/content/drive/MyDrive/tp_image/perroquet.jpeg'

img = cv2.imread(file_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Format BGR (OpenCV)')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.title('Format RGB (Matplotlib)')

plt.show()
```

```python
hauteur, largeur, canaux = img.shape
nombre_pixels = hauteur * largeur

print(f"Dimensions de l'image :")
print(f" - Hauteur  : {hauteur} pixels")
print(f" - Largeur  : {largeur} pixels")
print(f" - Canaux   : {canaux} (couleurs)")
print(f"Nombre total de pixels (hauteur x largeur) : {nombre_pixels} pixels")
```

```python

width = img.shape[1] // 2
height = img.shape[0] // 2
img_resized = cv2.resize(img, (width, height))
img_resized_rgb = cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB)

print(f"Dimensions d'origine : {img.shape[1]}x{img.shape[0]}")
print(f"Nouvelles dimensions : {width}x{height}")




img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)




plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(img_resized_rgb)
plt.title("Image redimensionnée (50%)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Image en niveaux de gris")
plt.axis("off")

plt.show()
```
