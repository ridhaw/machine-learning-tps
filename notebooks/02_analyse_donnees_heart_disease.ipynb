### Step 5– Class Balance

```python
# 1 & 2. Distribution de la variable cible et calcul des pourcentages
counts = df['target'].value_counts()
percentages = df['target'].value_counts(normalize=True) * 100

for val in counts.index:
    label = "Malade (target=1)" if val == 1 else "Sain (target=0)"
    print(f"{label} : {counts[val]} patients ({percentages[val]:.2f}%)")
```

```python
# 3. Création de DataFrames distincts et comparaison des moyennes
df_target_0 = df[df['target'] == 0]
df_target_1 = df[df['target'] == 1]

print("--- Moyennes pour les patients sains (target=0) ---")
display(df_target_0.mean())

print("\n--- Moyennes pour les patients malades (target=1) ---")
display(df_target_1.mean())
```

### Step 6– Missing and Invalid Values

```python
# 1. Compter le nombre de valeurs valant 0 dans les colonnes clés
columns_to_check = ['chol', 'trestbps', 'thalach']
zero_counts = (df[columns_to_check] == 0).sum()

print("Nombre de valeurs égales à 0 par colonne :")
print(zero_counts)
```

```python
import numpy as np

# 2. Remplacer les zéros invalides par NaN
df[columns_to_check] = df[columns_to_check].replace(0, np.nan)

# 3. Vérifier à nouveau les valeurs manquantes pour l'ensemble du dataset
print("Nombre total de valeurs manquantes (NaN) après remplacement :")
display(df.isnull().sum())
```

### Step 8– Grouped Analysis

```python
# 1. Définition de la fonction de calcul de moyenne groupée
def mean_target(var):
    """
    Calcule la moyenne d'une variable donnée, groupée par la colonne 'target'.
    """
    return df.groupby('target')[var].mean()
```

```python
# 2. Test de la fonction sur 'trestbps' et 'slope'
print("Moyennes de 'trestbps' groupées par target :")
display(mean_target('trestbps'))

print("\nMoyennes de 'slope' groupées par target :")
display(mean_target('slope'))
```
